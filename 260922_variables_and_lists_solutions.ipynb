````python
# 이 셀의 목적 — 문제 7-3. 빠진 IP를 채워 알림을 만든다
# 정답 — ip만 .get()으로 안전하게 꺼냅니다.

log = {"user": "kim01", "result": "FAIL", "time": "09:41"}
print(f"[{log['result']}] {log['user']} · IP {log.get('ip', '-')} · {log['time']}")
````
