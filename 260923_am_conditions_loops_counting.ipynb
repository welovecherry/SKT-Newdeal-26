### <mark style="display:block; background:#ffecb3; color:#1a1a1a; padding:6px 12px; border-radius:4px">1.1 실패 로그인지 확인하고 싶을 때</mark>

조건이 참일 때만 들여쓴 줄이 실행됩니다.


```python
# 로그인 로그 한 건을 준비합니다.
log = {"user": "kim01", "result": "FAIL", "fail_count": 2}

if log["result"] == "FAIL":  # 실패 로그인지 확인
    print("실패 로그입니다")  # 조건이 참일 때 실행
```

### <mark style="display:block; background:#ffecb3; color:#1a1a1a; padding:6px 12px; border-radius:4px">1.2 두 갈래로 나누고 싶을 때</mark>

`if`가 거짓이면 `else` 쪽이 실행됩니다.


```python
# 정상 로그인 로그 한 건을 준비합니다.
log = {"user": "lee02", "result": "OK", "fail_count": 0}

if log["result"] == "FAIL":  # 실패인지 확인
    print("확인 필요")  # 실패일 때 실행
else:  # 실패가 아닐 때 실행
    print("정상")
```

4. `else:` 아래에 네 칸 들여써서 「실패」를 출력합니다.

### <mark style="display:block; background:#ffecb3; color:#1a1a1a; padding:6px 12px; border-radius:4px">1.3 여러 단계로 나누고 싶을 때</mark>

위에서부터 검사하고 처음 참인 한 갈래만 실행됩니다.


```python
fail_count = 5  # 현재 실패 횟수

if fail_count >= 5:  # 5회 이상인지 확인
    print("높음")
elif fail_count >= 2:  # 2회 이상인지 확인
    print("주의")
else:  # 두 조건에 모두 해당하지 않을 때 실행
    print("낮음")
```

2. 위에서부터 검사해서 처음 참이 되는 갈래 하나만 실행됩니다.
3. 그래서 큰 기준(`>= 5`)을 먼저 검사합니다. `>= 2`를 먼저 쓰면 5회도 「주의」가 됩니다.
4. `>=`는 같은 값도 포함합니다. `fail_count`가 정확히 5여도 「높음」입니다.

## <mark style="display:block; background:#ffecb3; color:#1a1a1a; padding:6px 12px; border-radius:4px">2 · 반복문 — 같은 일을 모든 값에 적용한다</mark>

한 건에 하던 판단을 로그 목록 전체에 적용합니다.


<div style="background:#e8f5e9; color:#17351f; padding:12px 16px; border-radius:8px"><strong>🔎 코드를 읽는 법</strong><br>`for 값 in 목록`은 ‘목록에서 값을 하나씩 꺼내며’라고 읽습니다. 들여쓰기가 끝나는 곳까지가 한 번에 반복할 일입니다.</div>


### <mark style="display:block; background:#ffecb3; color:#1a1a1a; padding:6px 12px; border-radius:4px">2.1 리스트를 처음부터 끝까지 보고 싶을 때</mark>

`for`는 리스트에서 값을 하나씩 꺼냅니다.


```python
users = ["kim01", "lee02", "park03"]  # 확인할 계정 목록

for user in users:  # 계정을 하나씩 꺼내 반복
    print(user)
```

2. 줄 끝에 콜론(`:`)을 붙이고, 반복할 일은 네 칸 들여씁니다.
3. 반복할 일은 `print(user)` 한 줄입니다.

### <mark style="display:block; background:#ffecb3; color:#1a1a1a; padding:6px 12px; border-radius:4px">2.2 로그 딕셔너리를 하나씩 보고 싶을 때</mark>

한 번 돌 때마다 `log`에는 딕셔너리 하나가 들어갑니다.


```python
# 확인할 로그인 로그 세 건을 준비합니다.
logs = [
    {"user": "kim01", "result": "FAIL"},
    {"user": "lee02", "result": "OK"},
    {"user": "park03", "result": "FAIL"},
]

for log in logs:  # 로그를 하나씩 꺼내 반복
    print(log["user"], log["result"])
```

3. 결과는 `log["result"]`로, 계정은 `log["user"]`로 꺼냅니다.

### <mark style="display:block; background:#ffecb3; color:#1a1a1a; padding:6px 12px; border-radius:4px">2.3 실패한 계정만 모으고 싶을 때</mark>

빈 리스트를 만들고, 조건을 통과한 값만 `append()`합니다.


```python
# 확인할 로그인 로그 세 건을 준비합니다.
logs = [
    {"user": "kim01", "result": "FAIL"},
    {"user": "lee02", "result": "OK"},
    {"user": "park03", "result": "FAIL"},
]

failed_users = []  # 실패 계정을 담을 빈 목록
for log in logs:  # 로그를 하나씩 꺼내 반복
    if log["result"] == "FAIL":  # 실패 로그만 선택
        failed_users.append(log["user"])  # 계정을 목록에 추가

print(failed_users)
print(f"실패 {len(failed_users)}건")  # 목록 길이로 실패 건수 확인
```

<details>
<summary>▸ 힌트 보기</summary>

1. `failed_users = []`는 이미 있습니다. 빈 목록은 반복문 위에서 한 번만 만듭니다.
2. 문제 2-2처럼 `for` 안에 `if`를 넣어 실패 로그만 고릅니다.
3. `print` 대신 `failed_users.append(log["user"])`로 목록에 계정을 붙입니다.
4. 반복이 다 끝난 뒤 목록을 출력합니다. `print`는 `for`와 같은 칸(들여쓰기 없음)에 둡니다.

그래도 막히면 맨 아래 정답 2-3 셀의 「코드 표시」를 누릅니다.

### <mark style="display:block; background:#ffecb3; color:#1a1a1a; padding:6px 12px; border-radius:4px">3.3 계정과 횟수를 함께 돌고 싶을 때</mark>

`.items()`는 키와 값을 한 쌍씩 꺼냅니다.


```python
from collections import Counter  # 항목별 횟수를 세는 도구

# 실패 계정별 횟수를 계산합니다.
counts = Counter(["kim01", "kim01", "lee02", "park03", "kim01"])
for user, count in counts.items():  # 계정과 횟수를 함께 반복
    print(f"{user}: {count}회")
```

### <mark style="display:block; background:#ffecb3; color:#1a1a1a; padding:6px 12px; border-radius:4px">3.4 기준 이상인 계정만 찾고 싶을 때</mark>

기준값에는 대문자 이름을 붙여 한 곳에서 바꿉니다.


```python
from collections import Counter  # 항목별 횟수를 세는 도구

THRESHOLD = 2  # 확인이 필요한 기준 횟수
# 실패 계정별 횟수를 계산합니다.
counts = Counter(["kim01", "kim01", "lee02", "park03", "kim01"])

for user, count in counts.items():  # 계정과 횟수를 함께 반복
    if count >= THRESHOLD:  # 기준 이상인지 확인
        print(f"확인 필요: {user} · {count}회")
```
