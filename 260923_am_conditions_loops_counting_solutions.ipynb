#### 문제 3-4. 모든 횟수 출력하기

계정과 횟수를 「계정: N회」 모양으로 한 줄씩 출력하시오.

| | |
|---|---|
| 주어지는 값 | `counts` |
| 나와야 하는 결과 | `kim01: 2회`<br>`lee02: 1회` |


<div style="background:#bbdefb; color:#10233d; padding:12px 16px; border-radius:8px"><strong>✅ 정답 3-4</strong><br>집합은 중복을 지우고 Counter는 값별 횟수를 셉니다.</div>


```python
# 문제 3-4 정답
from collections import Counter
counts = Counter(["kim01", "kim01", "lee02"])
for user, count in counts.items():
    print(f"{user}: {count}회")
```

#### 문제 3-5. 기준 이상만 출력하기

실패가 2회 이상인 계정만 출력하시오.

| | |
|---|---|
| 주어지는 값 | `THRESHOLD = 2` · `counts` |
| 나와야 하는 결과 | `kim01` |


<div style="background:#bbdefb; color:#10233d; padding:12px 16px; border-radius:8px"><strong>✅ 정답 3-5</strong><br>집합은 중복을 지우고 Counter는 값별 횟수를 셉니다.</div>


```python
# 문제 3-5 정답
from collections import Counter
THRESHOLD = 2
counts = Counter(["kim01", "kim01", "lee02"])
for user, count in counts.items():
    if count >= THRESHOLD:
        print(user)
```

#### 문제 3-6. 기준 바꿔 다시 보기

기준을 3으로 바꾸고, 3회 이상인 계정과 횟수를 출력하시오.

| | |
|---|---|
| 주어지는 값 | 계정별 실패 횟수 |
| 나와야 하는 결과 | `park03 · 3회` |


<div style="background:#bbdefb; color:#10233d; padding:12px 16px; border-radius:8px"><strong>✅ 정답 3-6</strong><br>집합은 중복을 지우고 Counter는 값별 횟수를 셉니다.</div>


```python
# 문제 3-6 정답
from collections import Counter
THRESHOLD = 3
counts = Counter(["kim01", "kim01", "lee02", "park03", "park03", "park03"])
for user, count in counts.items():
    if count >= THRESHOLD:
        print(f"{user} · {count}회")
```

#### 문제 4-5. 확인할 계정 찾기

기준을 2회로 두고, 기준 이상인 계정을 「확인 필요: 계정 · N회」 모양으로 출력하시오.

| | |
|---|---|
| 주어지는 값 | `THRESHOLD = 2` |
| 나와야 하는 결과 | `확인 필요: kim01 · 3회` |


<div style="background:#bbdefb; color:#10233d; padding:12px 16px; border-radius:8px"><strong>✅ 정답 4-5</strong><br>오전의 조건·반복·집계 코드를 순서대로 연결합니다.</div>


```python
# 문제 4-5 정답
from collections import Counter
failed_users = ["kim01", "kim01", "kim01", "park03"]
THRESHOLD = 2
counts = Counter(failed_users)
for user, count in counts.items():
    if count >= THRESHOLD:
        print(f"확인 필요: {user} · {count}회")
```
