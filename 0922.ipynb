## <mark style="display:block; background:#ffecb3; color:#1a1a1a; padding:6px 12px; border-radius:4px">2 · 자료형 — 값의 갈래</mark>

값에도 갈래가 있습니다. 보안 로그에서 이 넷을 다 만납니다.

1. **int** — 소수점 없는 수 (실패 횟수 · 포트 번호)
2. **float** — 소수점 있는 수 (응답 시간 · 위험 점수)
3. **str** — 따옴표로 감싼 글자 (계정 이름 · IP 주소)
4. **bool** — `True` 아니면 `False` (차단 여부)

파이썬은 갈래를 미리 적지 않습니다. **넣는 값을 보고 알아서 정합니다.**

### <mark style="display:block; background:#ffecb3; color:#1a1a1a; padding:6px 12px; border-radius:4px">2.1 📖 예시 1 — 네 갈래를 만들어 본다</mark>

각각 하나씩 만들어 찍어 봅니다.

````python
# 이 셀의 목적 — 네 가지 자료형을 하나씩 만들어 본다
# bool 은 True / False 의 첫 글자가 대문자입니다. 소문자면 오류가 납니다.
fail_count = 3            # int
risk_score = 0.75         # float
user = "kim01"            # str
is_blocked = False        # bool
print(fail_count, risk_score, user, is_blocked)
````

### <mark style="display:block; background:#ffecb3; color:#1a1a1a; padding:6px 12px; border-radius:4px">2.2 📖 예시 2 — type() 으로 물어본다</mark>

헷갈리면 물어보면 됩니다.

````python
# 이 셀의 목적 — 값의 자료형을 type() 으로 확인한다
# class 뒤에 나오는 것이 자료형 이름입니다.
print(type(3))            # <class 'int'>
print(type("kim01"))      # <class 'str'>
print(type(False))        # <class 'bool'>
````

### <mark style="display:block; background:#ffecb3; color:#1a1a1a; padding:6px 12px; border-radius:4px">2.3 📖 예시 3 — 따옴표 하나가 갈라놓는다</mark>

## <mark style="display:block; background:#ffecb3; color:#1a1a1a; padding:6px 12px; border-radius:4px">3 · f-string — 문장 안에 값을 끼워 넣는다</mark>

오늘 배우는 것 중 **여덟 날 내내 제일 많이 쓸** 문법입니다.

1. 따옴표 **앞에 `f`** 를 붙입니다.
2. 값이 들어갈 자리에 **중괄호 `{}`** 로 이름을 적습니다.
3. 중괄호 안에서 **계산도 됩니다.**

### <mark style="display:block; background:#ffecb3; color:#1a1a1a; padding:6px 12px; border-radius:4px">3.1 📖 예시 1 — 가장 기본</mark>

이름 하나를 끼워 넣습니다.

````python
# 이 셀의 목적 — f-string 으로 문장 안에 값을 끼워 넣는다
# 따옴표 앞의 f 와 중괄호가 f-string 의 전부입니다.
user = "kim01"
print(f"{user}님이 로그인했습니다")
````

### <mark style="display:block; background:#ffecb3; color:#1a1a1a; padding:6px 12px; border-radius:4px">3.2 📖 예시 2 — 값 여러 개</mark>

중괄호를 여러 개 두면 됩니다. 숫자도 그냥 이름만 적습니다.

````python
# 이 셀의 목적 — 값을 여러 개 끼워 넣고, 중괄호 안에서 계산까지 해 본다
# 숫자를 넣을 때도 글자로 바꿀 필요가 없습니다. f-string 이 알아서 합니다.
user = "kim01"
fail_count = 5
print(f"{user} 계정이 {fail_count}회 실패했습니다")
print(f"한 번 더 실패하면 차단: {fail_count + 1}회")   # 중괄호 안에서 계산
````

## <mark style="display:block; background:#ffecb3; color:#1a1a1a; padding:6px 12px; border-radius:4px">4 · 형변환 — 값의 갈래를 바꾼다</mark>

**파일에서 읽은 값은 숫자처럼 보여도 전부 글자(str)입니다.** 실무에서 가장 자주 나는 오류가 여기서 납니다.

1. `int(x)` — 정수로 바꾼다
2. `str(x)` — 글자로 바꾼다
3. `float(x)` — 소수로 바꾼다

### <mark style="display:block; background:#ffecb3; color:#1a1a1a; padding:6px 12px; border-radius:4px">4.1 📖 예시 1 — 글자를 숫자로</mark>

int() 로 감싸면 더할 수 있게 됩니다.

## <mark style="display:block; background:#ffecb3; color:#1a1a1a; padding:6px 12px; border-radius:4px">5 · 리스트 — 여러 값을 순서대로 담는다</mark>

계정이 서른 개면 변수를 서른 개 만들 수 없습니다. 그래서 **한 이름에 여러 값을 담습니다.**

1. **대괄호 `[]`** 로 감싸고 **쉼표**로 나눕니다.
2. 꺼낼 때는 **번호(인덱스)** 로 꺼냅니다.
3. ⚠ **번호는 1 이 아니라 0 부터** 셉니다.

### <mark style="display:block; background:#ffecb3; color:#1a1a1a; padding:6px 12px; border-radius:4px">5.1 📖 예시 1 — 만들고 꺼낸다</mark>

번호가 0부터라는 것만 조심하면 됩니다.

````python
# 이 셀의 목적 — 리스트를 만들고 번호로 값을 꺼낸다
# 세 개짜리 리스트의 마지막은 [2] 입니다. [3] 을 부르면 IndexError 가 납니다.
users = ["kim01", "lee02", "park03"]
print(users[0])           # kim01  ← 첫 번째가 0번
print(users[2])           # park03 ← 마지막이 2번
print(len(users))         # 3      ← len = 개수 세기
````

### <mark style="display:block; background:#ffecb3; color:#1a1a1a; padding:6px 12px; border-radius:4px">5.2 📖 예시 2 — 나중에 더한다</mark>

### <mark style="display:block; background:#ffecb3; color:#1a1a1a; padding:6px 12px; border-radius:4px">5.3 📖 예시 3 — 실무에서 쓰는 모양</mark>

````python
# 이 셀의 목적 — 실패한 계정만 골라 리스트에 모으는 모양을 미리 본다
# 조건문·반복문은 Day 2 에 배웁니다. 지금은 모양만 보세요.
logs = ["kim01 FAIL", "lee02 OK", "park03 FAIL"]
failed_users = []
for line in logs:                    # 한 줄씩 꺼내서
    if "FAIL" in line:               # FAIL 이 들어 있으면
        failed_users.append(line)    # 모은다
print(failed_users)
print(f"실패 {len(failed_users)}건")
````

## <mark style="display:block; background:#ffecb3; color:#1a1a1a; padding:6px 12px; border-radius:4px">7 · .get() — 없는 키를 물어도 멈추지 않는다</mark>

**로그에는 빈 칸이 늘 있습니다.** 어떤 줄은 IP 가 있고 어떤 줄은 없습니다.

없는 키를 대괄호로 꺼내면 `KeyError` 가 나면서 **거기서 프로그램이 멈춥니다.** 수백 줄을 돌리다 한 줄에서 멈추면 나머지를 못 봅니다.

1. `log["ip"]` — 없으면 **멈춘다**. 반드시 있어야 하는 칸에 쓴다.
2. `log.get("ip")` — 없으면 `None`. 없을 수도 있는 칸에 쓴다.
3. `log.get("ip", "-")` — 없으면 `-`. 대신 쓸 값이 있을 때.

### <mark style="display:block; background:#ffecb3; color:#1a1a1a; padding:6px 12px; border-radius:4px">7.1 📖 예시 1 — 멈추는 쪽</mark>

일부러 에러를 내 봅니다. 빨간 글씨가 나오는 게 정상입니다.

````python
# 이 셀의 목적 — 없는 키를 대괄호로 꺼내면 어떻게 되는지 눈으로 본다
# 고칠 필요 없습니다. 그대로 실행해서 결과만 보세요.
log = {"user": "kim01", "result": "FAIL"}    # ip 가 없다
print(log["ip"])          # ← 여기서 멈춥니다 (KeyError)
print("검사 완료")         # 이 줄은 실행되지 않습니다
````

### <mark style="display:block; background:#ffecb3; color:#1a1a1a; padding:6px 12px; border-radius:4px">7.2 📖 예시 2 — 멈추지 않는 쪽</mark>

.get() 은 없으면 None 을 줍니다.

````python
# 이 셀의 목적 — .get() 이 없는 키를 물어도 멈추지 않는 것을 확인한다
# 맨 아래 "검사 완료" 가 찍히는지 보세요. 아까와 다릅니다.
log = {"user": "kim01", "result": "FAIL"}
print(log.get("ip"))                  # None = 값이 없다는 뜻
print(log.get("ip", "알 수 없음"))      # 두 번째 칸 = 없을 때 대신 쓸 값
print("검사 완료")                      # 이번엔 실행된다
````

### <mark style="display:block; background:#ffecb3; color:#1a1a1a; padding:6px 12px; border-radius:4px">7.3 📖 예시 3 — 실무에서 쓰는 모양</mark>

반드시 있어야 하는 칸과 없을 수도 있는 칸을 갈라 씁니다.

````python
# 이 셀의 목적 — 대괄호와 .get() 을 상황에 맞게 갈라 쓴다
# user 는 없으면 그게 문제다 → 대괄호. ip 는 없을 수 있다 → .get()
log = {"user": "kim01", "result": "FAIL"}
print(f"계정 {log['user']} · IP {log.get('ip', '-')} · 결과 {log['result']}")
````
