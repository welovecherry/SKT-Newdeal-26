### <mark style="display:block; background:#ffecb3; color:#1a1a1a; padding:6px 12px; border-radius:4px">5.1 여러 줄의 일에 이름을 붙이고 싶을 때</mark>

````python
def say_hello():  # say_hello라는 이름으로 적어 둔다 (아직 실행 안 됨)
    print("안녕하세요")

say_hello()  # 부르는 순간 실행된다
say_hello()  # 또 부르면 또 실행된다
````

> ⚠ 이렇게 쓰면 안 됩니다 — 이 코드는 실행하지 않습니다.

### <mark style="display:block; background:#ffecb3; color:#1a1a1a; padding:6px 12px; border-radius:4px">5.3 결과를 돌려받고 싶을 때</mark>

`return`은 함수가 만든 값을 부른 곳으로 돌려줍니다. 돌려받은 값은 변수에 담아 다음 일에 씁니다.

````python
def add(a, b):
    return a + b  # 결과를 부른 곳으로 돌려준다

result = add(3, 5)  # 돌려받은 8이 result에 담긴다
print(result)
````

> 🤔 이런 의문이 들 수 있습니다 — `print`와 `return`은 뭐가 다른가요?
>
> 1. `print`는 화면에 보여 주기만 합니다. 그 값을 변수에 담을 수 없습니다.
> 2. `return`은 값을 돌려줍니다. 받은 쪽에서 담아 두고 계산을 이어 갈 수 있습니다.
> 3. 그래서 결과를 다음 일에 쓸 함수는 `return`으로 끝냅니다.

````python
logs = [
    {"result": "FAIL"}, {"result": "OK"},
    {"result": "FAIL"}, {"result": "FAIL"},
]

def count_failed(logs):  # 오전 2-4의 코드를 함수로 묶었다
    fail_count = 0
    for log in logs:
        if log["result"] == "FAIL":
            fail_count = fail_count + 1
    return fail_count  # 반복이 끝난 뒤 돌려준다

total = count_failed(logs)
print(f"실패 {total}건")
````

### <mark style="display:block; background:#ffecb3; color:#1a1a1a; padding:6px 12px; border-radius:4px">5.4 값을 안 넘기면 정해 둔 값을 쓰고 싶을 때</mark>

매개변수에 `=값`을 붙이면 기본값이 됩니다. 넘기지 않으면 기본값을, 넘기면 넘긴 값을 씁니다.

````python
def needs_review(count, threshold=2):  # threshold를 안 넘기면 2
    return count >= threshold

print(needs_review(3))     # 기본값 2와 비교 → True
print(needs_review(3, 5))  # 넘긴 값 5와 비교 → False
````

### <mark style="display:block; background:#ffecb3; color:#1a1a1a; padding:6px 12px; border-radius:4px">5.5 오전 코드를 통째로 함수에 넣고 싶을 때</mark>

오전 3교시의 Counter와 기준값 코드를 함수로 묶으면, 어떤 목록이든 한 줄로 확인할 계정을 찾습니다.

````python
from collections import Counter  # 항목별 횟수를 세는 도구

def find_suspects(failed_users, threshold=2):
    counts = Counter(failed_users)
    suspects = []
    for user, count in counts.items():
        if count >= threshold:
            suspects.append(user)
    return suspects

print(find_suspects(["kim01", "kim01", "lee02", "kim01"]))
print(find_suspects(["kim01", "kim01", "lee02", "kim01"], 5))  # 기준을 바꿔 다시 확인
````

<div style="background:#e8f5e9; color:#17351f; padding:12px 16px; border-radius:8px"><strong>🔎 코드를 읽는 법</strong><br>파일 코드는 ‘연다 → 읽거나 쓴다 → 닫는다’ 순서입니다. `with` 아래 들여쓴 줄이 끝나면 파이썬이 알아서 닫아 줍니다. 만든 파일은 코랩 왼쪽 📁 파일 탭에서 볼 수 있습니다.</div>

### <mark style="display:block; background:#ffecb3; color:#1a1a1a; padding:6px 12px; border-radius:4px">6.1 파일에 글자를 쓰고 싶을 때</mark>

`"w"`는 쓰기(write) 모드입니다. 파일이 없으면 새로 만들고, 있으면 지우고 새로 씁니다.

````python
with open("memo.txt", "w", encoding="utf-8") as f:  # 쓰기 모드로 열고 f라고 부른다
    f.write("첫 번째 줄\n")  # \n은 줄바꿈 문자
    f.write("두 번째 줄\n")

print("memo.txt를 만들었습니다")
````

1. `encoding="utf-8"`은 한글이 깨지지 않게 하는 설정입니다. 파일을 열 때마다 붙입니다.
2. `write()`는 줄을 알아서 바꾸지 않습니다. 줄 끝마다 `\n`을 붙여야 다음 줄로 넘어갑니다.

### <mark style="display:block; background:#ffecb3; color:#1a1a1a; padding:6px 12px; border-radius:4px">7.1 자주 쓰는 함수를 따로 모아 두고 싶을 때</mark>

함수만 담은 파일을 모듈이라고 합니다. 저장하는 방법은 4.8과 6.6에서 쓴 `%%writefile` 그대로입니다.

````python
%%writefile utils.py
from collections import Counter

def find_suspects(failed_users, threshold=2):
    counts = Counter(failed_users)
    suspects = []
    for user, count in counts.items():
        if count >= threshold:
            suspects.append(user)
    return suspects
````

이제 `utils.py`가 생겼습니다. 다른 곳에서 이름만 불러 씁니다.

````python
from utils import find_suspects  # utils.py에서 find_suspects를 가져온다

print(find_suspects(["kim01", "kim01", "lee02", "kim01"]))
````

> ⚠ 이렇게 쓰면 안 됩니다 — 이 코드는 실행하지 않습니다.

> from utils import find_risky      # utils.py에 없는 이름
> ```
>
> ```
> ImportError: cannot import name 'find_risky' from 'utils'
> ```
>
> 모듈과 부르는 쪽을 잇는 것은 함수 이름 하나뿐입니다. `utils.py` 안에서 이름을 바꾸면 부르는 쪽도 같이 고쳐야 합니다.

### <mark style="display:block; background:#ffecb3; color:#1a1a1a; padding:6px 12px; border-radius:4px">7.4 글자로 적힌 시각에서 시만 꺼내고 싶을 때</mark>

`"03:11"`은 글자라서 시와 분을 따로 꺼낼 수 없습니다. `strptime`으로 시각 값으로 바꾸면 `.hour`로 시만 꺼냅니다.

````python
from datetime import datetime

time_text = "03:11"
log_time = datetime.strptime(time_text, "%H:%M")  # 글자를 시각 값으로

print(log_time.hour)    # 3
print(log_time.minute)  # 11
````

1. `strptime`의 두 번째 값은 「이 글자가 어떤 모양인지」 알려 주는 것입니다. `"03:11"`은 시와 분이라서 `"%H:%M"`입니다.
2. 모양이 다르면 오류가 납니다. `"09:01:30"`처럼 초까지 있으면 `"%H:%M:%S"`로 적습니다.

파이썬에는 이름 짓는 관례가 있습니다. PEP8이라고 부릅니다. 문법이 아니라서 어겨도 돌아가지만, 같이 일할 때 읽는 속도가 달라집니다.

| 무엇 | 어떻게 | 예 |
|---|---|---|
| 변수·함수 | 소문자와 밑줄 (snake_case) | `failed_users`, `count_failed` |
| 상수 (한 번 정하고 안 바꾸는 값) | 전부 대문자 | `THRESHOLD`, `API_TOKEN` |
| 이름 길이 | 줄인 이름보다 알아보는 이름 | `fc`보다 `fail_count` |

````python
fc = 3          # 무엇의 3인지 알 수 없다
fail_count = 3  # 실패 횟수라는 것이 보인다

print(fc, fail_count)
````

<table width="100%" bgcolor="#A5D6A7"><tr><td bgcolor="#A5D6A7"><h3><font color="#1B5E20">✍️ 문제 8-3 · 이름을 규칙에 맞게 고치기</font></h3></td></tr></table>

아래 코드는 잘 돌아가지만 이름이 규칙에 맞지 않습니다. 이름 세 개를 고쳐 같은 결과가 나오게 하시오.

| | |
|---|---|
| 주어지는 값 | `t` · `fu` · `CountFail` 을 쓰는 코드 |
| 🎯 나와야 하는 결과 | `확인 필요: kim01` |

<details>
<summary>▸ 힌트 보기</summary>

1. `t`는 한 번 정하고 안 바꾸는 기준값입니다. 상수라서 전부 대문자로 `THRESHOLD`가 좋습니다.
2. `fu`는 실패 계정 목록입니다. `failed_users`처럼 알아볼 수 있게 씁니다.
3. `CountFail`은 함수입니다. 소문자와 밑줄로 `count_fail`처럼 씁니다.
4. 이름을 바꿀 때는 그 이름을 쓰는 곳도 같이 바꿉니다. 한 군데만 바꾸면 `NameError`가 납니다.

그래도 막히면 맨 아래 정답 8-3 셀의 「코드 표시」를 누릅니다.

</details>

````python
t = 2
fu = ["kim01", "kim01", "lee02"]

def CountFail(users, name):
    count = 0
    for user in users:
        if user == name:
            count = count + 1
    return count

if CountFail(fu, "kim01") >= t:
    print("확인 필요: kim01")
````
