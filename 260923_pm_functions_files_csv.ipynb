### <mark style="display:block; background:#ffecb3; color:#1a1a1a; padding:6px 12px; border-radius:4px">5.5 오전 코드를 함수로 묶고 싶을 때</mark>

목록을 받아 확인할 계정 목록을 돌려줍니다.

```python
# 이 셀의 목적 — 실패 계정 목록을 받아 기준 이상인 계정을 돌려준다
# 필요한 도구를 파이썬에 가져옵니다.
from collections import Counter

# 아래 작업 묶음에 find_risky_users이라는 이름을 붙입니다.
def find_risky_users(failed_users, threshold=2):
    # 목록에 같은 값이 몇 번 나왔는지 셉니다.
    counts = Counter(failed_users)
    # 오른쪽에서 만든 값을 risky_users에 저장합니다.
    risky_users = []
    # 목록에서 값을 하나씩 꺼내 아래 들여쓴 작업을 반복합니다.
    for user, count in counts.items():
        # if 뒤의 조건이 True인지 확인합니다.
        if count >= threshold:
            # 조건에 맞는 값을 목록의 맨 뒤에 하나 추가합니다.
            risky_users.append(user)
    # 계산한 결과를 함수 밖으로 돌려줍니다.
    return risky_users

# 오른쪽에서 만든 값을 result에 저장합니다.
result = find_risky_users(["kim01", "kim01", "lee02", "kim01"])
# 현재 결과를 화면에 출력해 확인합니다.
print(result)
```

### <mark style="display:block; background:#ffecb3; color:#1a1a1a; padding:6px 12px; border-radius:4px">6.2 내 함수를 모듈로 옮기고 싶을 때</mark>

아래 셀은 연습용 파이썬 파일 하나를 만듭니다.

```python
%%writefile security_utils.py
# 필요한 도구를 파이썬에 가져옵니다.
from collections import Counter

# 아래 작업 묶음에 find_risky_users이라는 이름을 붙입니다.
def find_risky_users(failed_users, threshold=2):
    # 목록에 같은 값이 몇 번 나왔는지 셉니다.
    counts = Counter(failed_users)
    # 오른쪽에서 만든 값을 risky_users에 저장합니다.
    risky_users = []
    # 목록에서 값을 하나씩 꺼내 아래 들여쓴 작업을 반복합니다.
    for user, count in counts.items():
        # if 뒤의 조건이 True인지 확인합니다.
        if count >= threshold:
            # 조건에 맞는 값을 목록의 맨 뒤에 하나 추가합니다.
            risky_users.append(user)
    # 계산한 결과를 함수 밖으로 돌려줍니다.
    return risky_users
```

### <mark style="display:block; background:#ffecb3; color:#1a1a1a; padding:6px 12px; border-radius:4px">7.5 CSV 읽기를 함수로 묶고 싶을 때</mark>

파일 경로를 받아 실패 계정 목록을 돌려줍니다.

```python
# 이 셀의 목적 — CSV 경로를 받아 로그인 실패 계정을 돌려준다
# 필요한 도구를 파이썬에 가져옵니다.
import csv

# 아래 작업 묶음에 load_failed_users이라는 이름을 붙입니다.
def load_failed_users(filepath):
    # 오른쪽에서 만든 값을 failed_users에 저장합니다.
    failed_users = []
    # 파일을 열고, 들여쓴 작업이 끝나면 자동으로 닫습니다.
    with open(filepath, "r", encoding="utf-8") as file:
        # CSV의 첫 줄을 열 이름으로 읽는 도구를 준비합니다.
        reader = csv.DictReader(file)
        # 목록에서 값을 하나씩 꺼내 아래 들여쓴 작업을 반복합니다.
        for row in reader:
            # if 뒤의 조건이 True인지 확인합니다.
            if row["action"] == "LOGIN" and row["result"] == "FAIL":
                # 조건에 맞는 값을 목록의 맨 뒤에 하나 추가합니다.
                failed_users.append(row["user"])
    # 계산한 결과를 함수 밖으로 돌려줍니다.
    return failed_users

# 오른쪽에서 만든 값을 result에 저장합니다.
result = load_failed_users("login_logs.csv")
# 현재 결과를 화면에 출력해 확인합니다.
print(result)
```
