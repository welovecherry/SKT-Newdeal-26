#### 문제 5-5. 실패 계정 모으는 함수

로그 목록을 받아 실패한 계정 목록을 돌려주는 `collect_failed_users(logs)`를 만드시오.

| | |
|---|---|
| **주어지는 값** | 로그 딕셔너리 3건 |
| **나와야 하는 결과** | `['kim01', 'park03']` |


<div style="background:#bbdefb; color:#10233d; padding:12px 16px; border-radius:8px"><strong>✅ 정답 5-5</strong><br>여러 줄의 작업을 함수로 묶고 필요한 값은 매개변수로 받습니다.</div>


```python
# 문제 5-5 정답
def collect_failed_users(logs):
    failed_users = []
    for log in logs:
        if log["result"] == "FAIL":
            failed_users.append(log["user"])
    return failed_users

logs = [
    {"user": "kim01", "result": "FAIL"},
    {"user": "lee02", "result": "OK"},
    {"user": "park03", "result": "FAIL"},
]
print(collect_failed_users(logs))
```

#### 문제 5-6. 확인할 계정 찾는 함수

실패 계정 목록과 기준을 받아 기준 이상인 계정을 돌려주는 함수를 완성하시오.

| | |
|---|---|
| **주어지는 값** | 실패 계정 목록 · 기본 기준 2 |
| **나와야 하는 결과** | `['kim01']` |


<div style="background:#bbdefb; color:#10233d; padding:12px 16px; border-radius:8px"><strong>✅ 정답 5-6</strong><br>여러 줄의 작업을 함수로 묶고 필요한 값은 매개변수로 받습니다.</div>


```python
# 문제 5-6 정답
from collections import Counter

def find_risky_users(failed_users, threshold=2):
    counts = Counter(failed_users)
    risky_users = []
    for user, count in counts.items():
        if count >= threshold:
            risky_users.append(user)
    return risky_users

failed_users = ["kim01", "kim01", "lee02"]
print(find_risky_users(failed_users))
```

```python
# 문제 8-1 정답
import csv

def load_failed_users(filepath):
    failed_users = []
    with open(filepath, "r", encoding="utf-8") as file:
        reader = csv.DictReader(file)
        for row in reader:
            if row["action"] == "LOGIN" and row["result"] == "FAIL":
                failed_users.append(row["user"])
    return failed_users

print(load_failed_users("login_logs.csv"))
```

#### 문제 8-2. 기준 이상 계정 찾기

`find_risky_users(failed_users, threshold=2)`를 작성하시오.

| | |
|---|---|
| **주어지는 값** | 문제 8-1의 결과 |
| **나와야 하는 결과** | `['kim01']` |


<div style="background:#bbdefb; color:#10233d; padding:12px 16px; border-radius:8px"><strong>✅ 정답 8-2</strong><br>CSV 읽기와 계정 판정 함수를 차례로 연결합니다.</div>


```python
# 문제 8-2 정답
from collections import Counter

def find_risky_users(failed_users, threshold=2):
    counts = Counter(failed_users)
    risky_users = []
    for user, count in counts.items():
        if count >= threshold:
            risky_users.append(user)
    return risky_users

failed_users = ["kim01", "kim01", "kim01", "park03"]
print(find_risky_users(failed_users))
```
